```python
#dataframe
#create dataframe from dictionary of list
import pandas as pd
data={
    'name':['satwik','hari','soumesh'],
    'age':[25,30,45],
    'city':["jsg","banglore",'mumbai']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```
